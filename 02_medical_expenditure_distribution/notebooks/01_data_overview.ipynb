# Why Do Medical Expenditures Concentrate Among a Small Share of Individuals?


## STEP 1. Problem Framing

This project examines how medical expenditure concentration is associated with
age, reported health, income, inpatient utilization, services, and sources of
payment. The comparisons describe associations with spending concentration; they do
not identify causal mechanisms.

The mean describes spending per person and aggregate costs. The median,
quantiles, zero-spending share, and concentration shares describe other aspects
of the distribution, including how far a typical person's spending lies from
the mean. These summaries describe different aspects of expenditure. The project does not optimize predictions.


## STEP 2. Data Understanding — Variable Definition and Scope


### Dataset

- File: MEPS 2023 Full Year Consolidated File (HC-251).
- Unit: individual person; amounts are nominal 2023 U.S. dollars.
- Population estimates use positive annual person weights, `PERWT23F > 0`.
  Zero-weight records are counted separately and excluded from person-level estimates.
- The target is the represented U.S. civilian noninstitutionalized population,
  including people in scope for only part of the survey year. Their expenditures
  cover only the in-scope period; we do not annualize or remove them for this reason.
- Strata `VARSTR` and PSUs `VARPSU` are used for the overall mean's sampling uncertainty.

See [analysis definitions](../analysis_plan.md) for weight selection, boundary
rules, reserved-code handling, uncertainty, and source links.

## Target Variable Definition: `TOTEXP23`


### What `TOTEXP23` Measures
- **Concept:** Annual **total health care expenditures** for an individual in 2023.
- **Operational definition (MEPS):** Expenditures are defined as **payments** for health care services.
  - Specifically: **sum of direct payments** for care provided during the year.


### What Is Included (Payments)

`TOTEXP23` represents payments including:
- **Out-of-pocket** payments
- Payments by **private insurance**
- Payments by **public programs** such as **Medicaid and Medicare**
- Payments by other sources captured in MEPS “source of payment” categories

### What Is Not Included

- **Over-the-counter drugs** (not collected by MEPS)
- **Indirect payments not tied to specific medical events**, such as:
  - Medicaid Disproportionate Share payments
  - Medicare Direct Medical Education subsidies
- **Charges are not expenditures**:
  - Charges can differ substantially from actual payments due to discounts, bad debt, or free care.
  - Charge variables exist for some categories, but they do not represent actual dollars exchanged.


### How the Annual Total Is Constructed

- MEPS collects utilization and expenditures **at the event level** each round and **sums events into annual totals**.
- Expenditure data come from both:
  - **Household Component (HC)** and
  - **Medical Provider Component (MPC)**, where MPC payments are generally more accurate and are used to improve expenditure quality.
- Missing expenditures are addressed through **editing + imputation**:
  - Logical edits correct reporting issues (e.g., outliers, copays recorded as total payments, misclassified payer types)
  - Imputation includes **predictive mean matching (PMM)** and other procedures for capitated arrangements and flat/global fees.


### Structural Properties to Examine

`TOTEXP23` represents nonnegative payments. Zero totals can reflect no recorded
spending, including people with no use or services with no separate payment.
Zero spending is distinct from missing expenditure and zero survey weight.

Right skewness and concentrated spending are patterns to examine. They do not
by themselves establish a mathematical heavy-tail law or invalidate the mean.
We retain high expenditures, compare several distribution summaries, and use
`log1p` for a plot that includes zeros. Tail-law fitting is outside this analysis.


### Key Interpretation Note

- `TOTEXP23` reflects **realized spending/payments**, not “medical need.”
- Extremely high values are **not automatically treated as errors**; they may represent economically meaningful high-cost cases.

### Interpretation Guardrails

Interpret expenditure values using these definitions:
- This variable measures **realized spending**, not latent medical need
- Extreme values may correspond to economically meaningful high-cost individuals
- High-cost observations will not be treated as errors by default


### Analysis Design

The [expanded EDA](02_distribution_and_composition.ipynb) calculates weighted
means, quantiles, zero-spending shares, a Lorenz curve, Gini, and expenditure
concentration. It compares the top 5% with the population across age, reported
health, poverty category, and inpatient utilization, then decomposes expenditure
by service type and payer.

The [analysis definitions](../analysis_plan.md) fix the calculation rules and
interpretation limits. Top-percentile boundaries use proportional allocation
within spending ties. Only the overall mean has a survey-design confidence
interval; other comparisons are descriptive point estimates. Neither the
subgroup comparisons nor the spending decompositions establish causality.
